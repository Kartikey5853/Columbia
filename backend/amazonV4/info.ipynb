```python
import json
```

```python
with open("seller_products.json", encoding="utf8") as f:
    data = json.load(f)

main_products = sum(1 for p in data if p["is_main"])
variant_only = sum(1 for p in data if not p["is_main"])

print("Main products :", main_products)
print("Variant ASINs :", variant_only)
print("Total ASINs   :", len(data))
```

```python
from collections import Counter
```

```python

counter = Counter(len(p["family"]) for p in data if p["is_main"])

print(counter)
```
