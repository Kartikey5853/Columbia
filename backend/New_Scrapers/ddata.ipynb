```python
with open("ajio.json", "r", encoding="utf-8") as f:
    data = json.load(f)
```

```python
products = data["products"]
```

```python
from collections import defaultdict

groups = defaultdict(list)

for p in products:
    sku = p["sku"]
    prefix = "-".join(sku.split("-")[:-1])
    groups[prefix].append(sku)

for prefix, skus in groups.items():
    if len(skus) > 1:
        print(prefix, sorted(skus))
```

```python
[
    p for p in products
    if "BM6388-442" in p["sku"]
]
```

```python
codes = defaultdict(list)

for p in products:
    sku = p["sku"]
    prefix = "-".join(sku.split("-")[:-1])
    codes[prefix].append((p["product_id"], sku))

for prefix, vals in codes.items():
    if prefix == "BM6388-442":
        print(vals)
```

```python
def find(obj, target, path="root"):
    if isinstance(obj, dict):
        for k, v in obj.items():
            find(v, target, f"{path}.{k}")
    elif isinstance(obj, list):
        for i, v in enumerate(obj):
            find(v, target, f"{path}[{i}]")
    elif isinstance(obj, str):
        if target in obj:
            print(path, "=", obj)

find(raw, "BM6388-442")
```
